```python
import pandas as pd

# 1. Read the hacker_news.csv file from data directory
file_path = 'data/hacker_news.csv'
df = pd.read_csv(file_path)
```

```python
# # 2. Get the first five rows
print("--- First 5 Rows ---")
print(df.head())
```

```python
# # 3. Get the last five rows
print("\n--- Last 5 Rows ---")
print(df.tail())
```

```python
# # 4. Get the title column as pandas series
titles_series = df['title']
print("\n--- Title Column (First 5 entries as Series) ---")
print(titles_series.head())
```

```python
# # 5. Count the number of rows and columns
num_rows, num_cols = df.shape
print(f"\n--- Shape of Data ---")
print(f"Number of rows: {num_rows}")
print(f"Number of columns: {num_cols}")
```

```python
# # - Filter the titles which contain python (case-insensitive)
python_titles = df[df['title'].str.contains('python', case=False, na=False)]
print(f"\n--- Titles containing 'python' (Count: {len(python_titles)}) ---")
print(python_titles['title'].head())
```

```python
# # - Filter the titles which contain JavaScript (case-insensitive)
js_titles = df[df['title'].str.contains('javascript', case=False, na=False)]
print(f"\n--- Titles containing 'JavaScript' (Count: {len(js_titles)}) ---")
print(js_titles['title'].head())
```

```python
# # - Explore the data and make sense of it
print("\n--- Basic Data Exploration ---")
print(df.info())
print("\nSummary Statistics:")
print(df.describe(include='all'))
```
