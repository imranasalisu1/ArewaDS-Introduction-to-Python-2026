```python
# 1. Scrape Boston University Facts & Stats and Save as JSON
import json
import requests
from bs4 import BeautifulSoup

def scrape_bu_facts(url='https://www.bu.edu/president/boston-university-facts-stats/'):
    headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64)'}
    response = requests.get(url, headers=headers)

    if response.status_code != 200:
        print(f"Failed to fetch page. Status code: {response.status_code}")
        return

    soup = BeautifulSoup(response.text, 'html.parser')

    # Extract headers and paragraphs or fact blocks
    data = []
    # Adjust selectors based on the site's structure (e.g., headings and content blocks)
    for section in soup.find_all(['h2', 'h3', 'p']):
        text = section.get_text(strip=True)
        if text:
            data.append({'tag': section.name, 'text': text})

    # Save to JSON file
    with open('bu_facts.json', 'w', encoding='utf-8') as f:
        json.dump(data, f, indent=4, ensure_ascii=False)
    print("Successfully scraped BU facts and saved to 'bu_facts.json'.")

# Run the function
scrape_bu_facts()
```

```python
# 2. Extract Table from UCI ML Repository and Save as JSON
import pandas as pd

def scrape_uci_datasets(url='https://archive.ics.uci.edu/ml/datasets.php'):
    try:
        # pd.read_html looks for HTML <table> tags and returns a list of DataFrames
        tables = pd.read_html(url)
        if tables:
            df = tables[0]
            # Convert DataFrame to JSON format
            df.to_json('uci_datasets.json', orient='records', indent=4)
            print("Successfully extracted UCI table and saved to 'uci_datasets.json'.")
        else:
            print("No tables found on the target page.")
    except Exception as e:
        print(f"An error occurred: {e}")

# Run the function
# scrape_uci_datasets()
```

```python
# 3. Scrape US Presidents Wikipedia Table and Save as JSON
import pandas as pd

def scrape_us_presidents(url='https://en.wikipedia.org/wiki/List_of_presidents_of_the_United_States'):
    try:
        print("Scraping Wikipedia table (this may take a few moments)...")
        tables = pd.read_html(url)

        # Typically the main chronological list of presidents is the primary table on this page
        # We can select the table containing key headers like 'President' or 'Term'
        presidents_df = None
        for table in tables:
            if any('President' in str(col) for col in table.columns):
                presidents_df = table
                break

        if presidents_df is not None:
            # Clean up multi-index columns if present
            if isinstance(presidents_df.columns, pd.MultiIndex):
                presidents_df.columns = ['_'.join(col).strip() for col in presidents_df.columns.values]

            # Export to JSON
            presidents_df.to_json('us_presidents.json', orient='records', indent=4)
            print("Successfully scraped U.S. Presidents table and saved to 'us_presidents.json'.")
        else:
            print("Could not locate the presidents table automatically.")
    except Exception as e:
        print(f"An error occurred during scraping: {e}")

# Run the function
scrape_us_presidents()
```
