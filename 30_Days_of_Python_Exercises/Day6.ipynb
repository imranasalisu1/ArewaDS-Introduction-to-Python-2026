```python
# 1. Create an empty dictionary called dog
dog = {}
print(dog)
```

```python
# 2. Add name, color, breed, legs, age to the dog dictionary
dog['name'] = 'Buddy'
dog['color'] = 'Brown'
dog['breed'] = 'Labrador'
dog['legs'] = 4
dog['age'] = 3
print(dog)
```

```python
# 3. Create a student dictionary with the required keys
student = {
    'first_name': 'Asabeneh',
    'last_name': 'Yetayeh',
    'gender': 'Male',
    'age': 250,
    'marital_status': 'Married',
    'skills': ['Python', 'JavaScript', 'React', 'Node', 'MongoDB'],
    'country': 'Finland',
    'city': 'Helsinki',
    'address': 'Espoo, Finland'
}
print(student)
```

```python
# 4. Get the length of the student dictionary
print(len(student))
```

```python
# 5. Get the value of skills and check its data type
print(student['skills'])
print(type(student['skills']))
print(isinstance(student['skills'], list))
```

```python
# 6. Modify the skills values by adding one or two skills
student['skills'].append('HTML')
student['skills'].append('CSS')
print(student['skills'])
```

```python
# 7. Get the dictionary keys as a list
keys = list(student.keys())
print(keys)
```

```python
# 8. Get the dictionary values as a list
values = list(student.values())
print(values)
```

```python
# 9. Change the dictionary to a list of tuples using items()
student_items = list(student.items())
print(student_items)
```

```python
# 10. Delete one of the items in the dictionary
student.pop('address')          # removes 'address' key
# or: del student['address']
print(student)
```
