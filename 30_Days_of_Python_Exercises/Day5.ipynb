**Sets Exercises**

```python
# Initial data
it_companies = {'Facebook', 'Google', 'Microsoft', 'Apple', 'IBM', 'Oracle', 'Amazon'}
A = {19, 22, 24, 20, 25, 26}
B = {19, 22, 20, 25, 26, 24, 28, 27}
age = [22, 19, 24, 25, 26, 24, 25, 24]
```

```python
# Sets Exercises: Level 1
# 1. Find the length of the set it_companies
print(len(it_companies))
```

```python
# 2. Add 'Twitter' to it_companies
it_companies.add('Twitter')
print(it_companies)
```

```python
# 3. Insert multiple IT companies at once
it_companies.update(['LinkedIn', 'Tesla', 'Netflix'])
print(it_companies)
```

```python
# 4. Remove one of the companies from the set
it_companies.remove('Tesla')
print(it_companies)
```

```python
# 5. Difference between remove and discard
"""
remove():
  - Removes the specified element from the set.
  - Raises a KeyError if the element is NOT present.

discard():
  - Removes the specified element from the set.
  - Does NOT raise an error if the element is NOT present (silently does nothing).
"""
# Example:
it_companies.remove('Amazon')
# it_companies.remove('Amazon')
it_companies.discard('Amazon')
```

**Sets Exercises: Level 2**

```python
# 6. Join A and B
print(A.union(B))
```

```python
# 7. Find A intersection B
print(A.intersection(B))
```

```python
# 8. Is A subset of B?
print(A.issubset(B))
```

```python
# 9. Are A and B disjoint sets?
print(A.isdisjoint(B))
```

```python
# 10. Join A with B and B with A
print(A.union(B))   # A joined with B
print(B.union(A))   # B joined with A
# Both produce the same result: {19, 20, 22, 24, 25, 26, 27, 28}
```

```python
# 11. Symmetric difference between A and B
print(A.symmetric_difference(B))
```

```python
# 12. Delete the sets completely
del A
del B
```

**Sets Exercises: Level 3**

```python
# 13. Convert ages to a set and compare lengths
age_set = set(age)
print('List length:', len(age))
print('Set length:', len(age_set))
print('List is bigger:', len(age) > len(age_set))
```

```python
# 14. Difference between string, list, tuple, and set
"""
STRING:
  - Ordered sequence of characters.
  - Immutable (cannot be changed after creation).
  - Allows duplicates.
  - Indexed and sliceable.
  - Example: 'hello'

LIST:
  - Ordered, mutable collection of items.
  - Can hold mixed data types.
  - Allows duplicates.
  - Indexed and sliceable.
  - Example: [1, 2, 'a', 3.5]

TUPLE:
  - Ordered, immutable collection of items.
  - Can hold mixed data types.
  - Allows duplicates.
  - Indexed and sliceable.
  - Example: (1, 2, 'a', 3.5)

SET:
  - Unordered, mutable collection of unique items.
  - Does NOT allow duplicates.
  - NOT indexed or sliceable.
  - Supports mathematical set operations (union, intersection, etc.).
  - Example: {1, 2, 3}
"""
```

```python
# 15. Count unique words in a sentence
sentence = 'I am a teacher and I love to inspire and teach people'
words = sentence.split()
unique_words = set(words)
print('Number of unique words:', len(unique_words))
```
