**Random Module Exercises**

```python
# Exercises: Level 1
# random_user_id
import random
import string
def random_user_id():
    chars = string.ascii_letters + string.digits  # a-z, A-Z, 0-9
    return ''.join(random.choice(chars) for _ in range(6))

print(random_user_id())
```

```python
# 2. user_id_gen_by_user
def user_id_gen_by_user():
    length = int(input("Enter number of characters: "))
    count = int(input("Enter number of IDs to generate: "))
    chars = string.ascii_letters + string.digits

    ids = []
    for _ in range(count):
        ids.append(''.join(random.choice(chars) for _ in range(length)))
    return '\n'.join(ids)

print(user_id_gen_by_user())
# Input: 5 5

print(user_id_gen_by_user())
# Input: 16 5
```

```python
# 3. rgb_color_gen
def rgb_color_gen():
    r = random.randint(0, 255)
    g = random.randint(0, 255)
    b = random.randint(0, 255)
    return f"rgb({r},{g},{b})"

print(rgb_color_gen())
```

**Exercises: Level 2**

```python
# 4. list_of_hexa_colors
def list_of_hexa_colors(n=1):
    hex_digits = '0123456789abcdef'
    colors = []
    for _ in range(n):
        color = '#' + ''.join(random.choice(hex_digits) for _ in range(6))
        colors.append(color)
    return colors

print(list_of_hexa_colors(3))
print(list_of_hexa_colors(1))
```

```python
# 5. list_of_rgb_colors
def list_of_rgb_colors(n=1):
    colors = []
    for _ in range(n):
        r = random.randint(0, 255)
        g = random.randint(0, 255)
        b = random.randint(0, 255)
        colors.append(f"rgb({r},{g},{b})")
    return colors

print(list_of_rgb_colors(3))
```

```python
# 6. generate_colors
def generate_colors(color_type, n=1):
    if color_type == 'hexa':
        return list_of_hexa_colors(n)
    elif color_type == 'rgb':
        return list_of_rgb_colors(n)
    else:
        return "Invalid color type. Use 'hexa' or 'rgb'."

print(generate_colors('hexa', 3))
print(generate_colors('hexa', 1))
print(generate_colors('rgb', 3))
print(generate_colors('rgb', 1))
```

**Exercises: Level 3**

```python
# 7. shuffle_list
def shuffle_list(lst):
    shuffled = lst.copy()
    random.shuffle(shuffled)
    return shuffled

original = [1, 2, 3, 4, 5]
print(shuffle_list(original))
print(original)
```

```python
# 8. Seven unique random numbers (0–9)
def seven_unique_numbers():
    return random.sample(range(10), 7)

print(seven_unique_numbers())
```
