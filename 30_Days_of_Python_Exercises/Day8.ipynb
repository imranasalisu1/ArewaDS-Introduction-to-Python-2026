**Exercises: Level 1**

```python
# 1. Iterate 0 to 10 using for loop
print("For loop 0 to 10:")
for i in range(11):
    print(i)
```

```python
# Same using while loop
print("While loop 0 to 10:")
i = 0
while i <= 10:
    print(i)
    i += 1
```

```python
# 2. Iterate 10 to 0 using for loop
print("For loop 10 to 0:")
for i in range(10, -1, -1):
    print(i)
```

```python
# Same using while loop
print("While loop 10 to 0:")
i = 10
while i >= 0:
    print(i)
    i -= 1
```

```python
# 3. Triangle pattern using 7 calls to print()
for i in range(1, 8):
    print('#' * i)
```

```python
# 4. 8x8 grid of "# " using nested loops
for row in range(8):
    for col in range(8):
        print('#', end=' ')
    print()
```

```python
# 5. Multiplication table 0x0 to 10x10
for i in range(11):
    print(f"{i} x {i} = {i * i}")
```

```python
# 6. Iterate through the list
lst = ['Python', 'Numpy', 'Pandas', 'Django', 'Flask']
for item in lst:
    print(item)
```

```python
# 7. Print only even numbers from 0 to 100
print("Even numbers:")
for i in range(0, 101, 2):
    print(i)
```

```python
# 8. Print only odd numbers from 0 to 100
print("Odd numbers:")
for i in range(1, 101, 2):
    print(i)
```

**Exercises: Level 2**

```python
# 9. Sum of all numbers from 0 to 100
total = 0
for i in range(101):
    total += i
print(f"The sum of all numbers is {total}.")
```

```python
# 10. Sum of all evens and odds from 0 to 100
sum_even = 0
sum_odd = 0
for i in range(101):
    if i % 2 == 0:
        sum_even += i
    else:
        sum_odd += i

print(f"The sum of all evens is {sum_even}. And the sum of all odds is {sum_odd}.")
```

```python
# Total number of unique languages
all_languages = set()
for country in countries_data:
    for lang in country['languages']:
        all_languages.add(lang)
print("Total number of unique languages:", len(all_languages))

# Ten most spoken languages (count = number of countries listing the language)
language_count = {}
for country in countries_data:
    for lang in country['languages']:
        language_count[lang] = language_count.get(lang, 0) + 1

top_languages = sorted(language_count.items(), key=lambda x: x[1], reverse=True)[:10]
print("\n10 Most Spoken Languages:")
for lang, count in top_languages:
    print(f"{lang}: {count}")

# 10 most populated countries
top_populated = sorted(countries_data, key=lambda c: c['population'], reverse=True)[:10]
print("\n10 Most Populated Countries:")
for country in top_populated:
    print(f"{country['name']}: {country['population']:,}")
```
