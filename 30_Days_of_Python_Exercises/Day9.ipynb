```python
import math
from statistics import median, mode, variance, pstdev
```

```python
# 1. add_two_numbers
def add_two_numbers(a, b):
    return a + b
print(add_two_numbers(3, 5))
```

```python
# 2. area_of_circle
def area_of_circle(r):
    return math.pi * r * r
print(area_of_circle(10))
```

```python
# 3. add_all_nums (arbitrary args + type check)
def add_all_nums(*args):
    if not all(isinstance(x, (int, float)) for x in args):
        return "All items must be numbers."
    return sum(args)
print(add_all_nums(1, 2, 3, 4))
print(add_all_nums(1, 'a', 3))
```

```python
# 4. convert_celsius_to_fahrenheit
def convert_celsius_to_fahrenheit(c):
    return (c * 9/5) + 32
print(convert_celsius_to_fahrenheit(0))
print(convert_celsius_to_fahrenheit(100))
```

```python
# 5. check_season
def check_season(month):
    month = month.strip().capitalize()
    if month in ('September', 'October', 'November'):
        return 'Autumn'
    elif month in ('December', 'January', 'February'):
        return 'Winter'
    elif month in ('March', 'April', 'May'):
        return 'Spring'
    elif month in ('June', 'July', 'August'):
        return 'Summer'
    return 'Invalid month'
print(check_season('October'))
```

```python
# 6. calculate_slope (slope of line through two points)
def calculate_slope(x1, y1, x2, y2):
    if x2 == x1:
        return 'Undefined (vertical line)'
    return (y2 - y1) / (x2 - x1)
print(calculate_slope(1, 2, 3, 6))
```

```python
# 7. solve_quadratic_eqn: ax² + bx + c = 0
def solve_quadratic_eqn(a, b, c):
    if a == 0:
        return 'Not a quadratic equation (a=0)'
    d = b**2 - 4*a*c
    if d > 0:
        r1 = (-b + math.sqrt(d)) / (2*a)
        r2 = (-b - math.sqrt(d)) / (2*a)
        return (r1, r2)
    elif d == 0:
        return (-b / (2*a),)
    else:
        real = -b / (2*a)
        imag = math.sqrt(-d) / (2*a)
        return (complex(real, imag), complex(real, -imag))
print(solve_quadratic_eqn(1, -3, 2))
```

```python
# 8. print_list
def print_list(lst):
    for item in lst:
        print(item)
print_list(['a', 'b', 'c'])
```

```python
# 9. reverse_list (using loops)
def reverse_list(lst):
    result = []
    for i in range(len(lst) - 1, -1, -1):
        result.append(lst[i])
    return result
print(reverse_list([1, 2, 3, 4, 5]))
print(reverse_list(["A", "B", "C"]))
```

```python
# 10. capitalize_list_items
def capitalize_list_items(lst):
    return [str(item).capitalize() for item in lst]
print(capitalize_list_items(['potato', 'tomato']))
```

```python
# 11. add_item
def add_item(lst, item):
    return lst + [item]
food_staff = ['Potato', 'Tomato', 'Mango', 'Milk']
print(add_item(food_staff, 'Meat'))
# ['Potato', 'Tomato', 'Mango', 'Milk', 'Meat']
print(add_item([2, 3, 7, 9], 5))
```

```python
# 12. remove_item
def remove_item(lst, item):
    result = lst.copy()
    if item in result:
        result.remove(item)
    return result
food_staff = ['Potato', 'Tomato', 'Mango', 'Milk']
print(remove_item(food_staff, 'Mango'))
print(remove_item([2, 3, 7, 9], 3))
```

```python
# 13. sum_of_numbers
def sum_of_numbers(n):
    return sum(range(1, n + 1))
print(sum_of_numbers(5))
print(sum_of_numbers(10))
print(sum_of_numbers(100))
```

```python
# 14. sum_of_odds
def sum_of_odds(n):
    return sum(i for i in range(1, n + 1) if i % 2 != 0)
print(sum_of_odds(10))
```

```python
# 15. sum_of_even
def sum_of_even(n):
    return sum(i for i in range(1, n + 1) if i % 2 == 0)
print(sum_of_even(10))
```

```python
# 16. evens_and_odds
def evens_and_odds(n):
    evens = sum(1 for i in range(n + 1) if i % 2 == 0)
    odds = (n + 1) - evens
    print(f"The number of odds are {odds}.")
    print(f"The number of evens are {evens}.")
evens_and_odds(100)
```

```python
# 17. factorial
def factorial(n):
    if n < 0:
        return 'Factorial not defined for negatives'
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result
print(factorial(5))
```

```python
# 18. is_empty
def is_empty(value):
    try:
        return len(value) == 0
    except TypeError:
        return True
print(is_empty([]))
print(is_empty([1]))
print(is_empty(''))
print(is_empty({}))
```

```python
# 19. Statistics functions
def calculate_mean(lst):
    return sum(lst) / len(lst) if lst else None

def calculate_median(lst):
    return median(lst) if lst else None

def calculate_mode(lst):
    try:
        return mode(lst)
    except Exception:
        return None

def calculate_range(lst):
    return max(lst) - min(lst) if lst else None

def calculate_variance(lst):
    return variance(lst) if len(lst) > 1 else 0

def calculate_std(lst):
    return pstdev(lst) if lst else None

data = [1, 2, 2, 3, 4, 5]
print(calculate_mean(data))
print(calculate_median(data))
print(calculate_mode(data))
print(calculate_range(data))
print(calculate_variance(data))
print(calculate_std(data))
```

```python
# 20. is_prime
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
print(is_prime(11))
print(is_prime(12))
```

```python
# 21. Check all items unique
def all_unique(lst):
    return len(lst) == len(set(lst))
print(all_unique([1, 2, 3]))
print(all_unique([1, 2, 2, 3]))
```

```python
# 22. Check all items same data type
def all_same_type(lst):
    if not lst:
        return True
    first_type = type(lst[0])
    return all(type(x) == first_type for x in lst)
print(all_same_type([1, 2, 3]))
print(all_same_type([1, 'a', 3]))
```

```python
# 23. Check valid Python variable name
def is_valid_variable(name):
    return isinstance(name, str) and name.isidentifier() and not name[0].isdigit()
print(is_valid_variable('my_var'))
print(is_valid_variable('30days'))
print(is_valid_variable('my-var'))
```

```python
# 24. most_spoken_languages and most_populated_countries
# Using the countries_data structure (list of dicts with 'name', 'languages', 'population')
def most_spoken_languages(countries_data, top=10):
    counts = {}
    for c in countries_data:
        for lang in c['languages']:
            counts[lang] = counts.get(lang, 0) + 1
    return sorted(counts.items(), key=lambda x: x[1], reverse=True)[:top]

def most_populated_countries(countries_data, top=10):
    return sorted(countries_data, key=lambda c: c['population'], reverse=True)[:top]

# Example with inline data (replace with real countries_data when available):
sample = [
    {'name': 'China', 'languages': ['Mandarin'], 'population': 1400050000},
    {'name': 'India', 'languages': ['Hindi', 'English'], 'population': 1350000000},
    {'name': 'USA', 'languages': ['English'], 'population': 330000000},
    {'name': 'Canada', 'languages': ['English', 'French'], 'population': 38000000},
]
print(most_spoken_languages(sample, 3))
# [('English', 3), ('Hindi', 1), ('Mandarin', 1)]
print([c['name'] for c in most_populated_countries(sample, 3)])
# ['China', 'India', 'USA']
```

**List Comprehensions Exercises**

```python
# 1. Filter only negative and zero
numbers = [-4, -3, -2, -1, 0, 2, 4, 6]
filtered = [n for n in numbers if n <= 0]
print(filtered)

# 2. Flatten 3-level nested list
list_of_lists = [[[1, 2, 3]], [[4, 5, 6]], [[7, 8, 9]]]
flat = [n for sub1 in list_of_lists for sub2 in sub1 for n in sub2]
print(flat)

# 3. List of tuples (powers)
result = [(i, 1, i, i**2, i**3, i**4, i**5) for i in range(11)]
for row in result:
    print(row)

# 4. Flatten countries with uppercase + 3-letter code
countries = [[('Finland', 'Helsinki')], [('Sweden', 'Stockholm')], [('Norway', 'Oslo')]]
output = [
    [country.upper(), country.upper()[:3], city.upper()]
    for group in countries
    for (country, city) in group
]
print(output)

# 5. List of dictionaries
countries = [[('Finland', 'Helsinki')], [('Sweden', 'Stockholm')], [('Norway', 'Oslo')]]
dicts = [
    {'country': country.upper(), 'city': city.upper()}
    for group in countries
    for (country, city) in group
]
print(dicts)

# 6. List of concatenated strings
names = [[('Asabeneh', 'Yetayeh')], [('David', 'Smith')],
         [('Donald', 'Trump')], [('Bill', 'Gates')]]
full_names = [f"{first} {last}" for group in names for (first, last) in group]
print(full_names)

# 7. Lambda for slope and y-intercept of a linear function
# Line: y = mx + b, given two points (x1, y1), (x2, y2)
slope = lambda x1, y1, x2, y2: (y2 - y1) / (x2 - x1) if x2 != x1 else None
y_intercept = lambda x1, y1, x2, y2: y1 - slope(x1, y1, x2, y2) * x1

print(slope(1, 2, 3, 6))
print(y_intercept(1, 2, 3, 6))
# Line: y = 2x + 0
```
