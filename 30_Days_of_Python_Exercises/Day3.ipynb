```python
#8. Calculate the slope, x-intercept and y-intercept of y = 2x -2
# Line equation: y = mx + b, where m is slope and b is y-intercept
# Given: y = 2x - 2

m = 2      # slope
b = -2     # y-intercept

print("=== Line: y = 2x - 2 ===")
print(f"Slope (m)     = {m}")
print(f"Y-intercept   = {b}  -> point (0, {b})")

# x-intercept: set y = 0 and solve for x -> 0 = 2x - 2 -> x = 1
x_intercept = -b / m
print(f"X-intercept   = {x_intercept}  -> point ({x_intercept}, 0)")
```

```python
#9. Slope is (m = y2-y1/x2-x1). Find the slope and Euclidean distance between point (2, 2) and point (6,10)
import math

# Points
x1, y1 = 2, 2
x2, y2 = 6, 10

# Slope: m = (y2 - y1) / (x2 - x1)
slope = (y2 - y1) / (x2 - x1)

# Euclidean distance: d = sqrt((x2 - x1)^2 + (y2 - y1)^2)
distance = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

print("=== Points: (2, 2) and (6, 10) ===")
print(f"Slope (m)              = {slope}")
print(f"Euclidean distance (d) = {distance}")
```

```python
#11. Calculate the value of y (y = x^2 + 6x + 9). Try to use different x values and figure out at what x value y is going to be 0.
def y_func(x):
    return x**2 + 6*x + 9

print("=== y = x^2 + 6x + 9 ===")
print(f"{'x':>6} | {'y':>8}")
print("-" * 18)

for x in range(-8, 1):          # try x = -8 .. 0
    print(f"{x:>6} | {y_func(x):>8}")

# Find x where y == 0
roots = [x for x in range(-10, 5) if y_func(x) == 0]
print(f"\nFound root(s): x = {roots}")
```
