```python
temperatures = [29, 36, 41, 33, 38, 27, 40]     # one reading per day, in degrees

hotDays = 0                    # counter                   # list accumulator
for temp in temperatures:
    if temp >= 38:                         # FILTER: only some days pass this test
        hotDays += 1
        hotList.append(temp)
        print(temp, "-> hot")
    elif temp >= 30:                       # CLASSIFY: every day gets a label
        print(temp, "-> warm")
    else:
        print(temp, "-> mild")

print("Hot days:", hotDays)
print("Their temperatures:", hotList)
```

```python
# Saving up for a phone. We do NOT know in advance how many months it takes.
PHONE_PRICE = 95000
MONTHLY_SAVING = 12000

savings = 0                          # 1. initialise
months = 0

while savings < PHONE_PRICE:         # 2. test: keep going as long as we do not have enough
    savings += MONTHLY_SAVING        # 3. update: this is what will make the condition False one day
    months += 1
    print("Month", months, "- saved so far:", savings)

print("You can buy the phone after", months, "months.")
print("Money left over:", savings - PHONE_PRICE)
```

```python
loan = 384000
monthly_deduction = 32000
savings = 0
months = 0
while savings < loan:
    savings = savings + monthly_deduction
    months = months + 1
    print("months",months,"-saved so far:",savings)
print("your loan is over after",months)
print("what left after deduction:",loan - savings)
```
