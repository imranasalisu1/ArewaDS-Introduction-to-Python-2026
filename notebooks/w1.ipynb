## 4️⃣ Type Conversion: `int()` and `float()`

Remember: `input()` always returns a **string**. When you need to do math with numbers from user input, you must convert them!

- `int()` - converts to an integer (whole number)
- `float()` - converts to a floating-point number (decimal number)

### Syntax:
```python
number = int(input("Enter a number: "))
decimal = float(input("Enter a decimal: "))
```

### Example 4.1: Why Type Conversion Matters

````python
# Without conversion - this will concatenate (join) strings!
num1 = input("Enter first number: ")
num2 = input("Enter second number: ")
result = num1 + num2
print("Result:", result)  # Try entering 5 and 3 - you'll get "53" instead of 8!
````

````python
# With int() conversion - now it does math!
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
result = num1 + num2
print("Result:", result)  # Now 5 and 3 will give you 8!
````

### Example 4.2: Adding Two Numbers (from Class)

````python
# This is the example from our slides!
first = int(input("Enter the first number: "))
second = int(input("Enter the second number: "))
print("The sum is", first + second)
````

### Example 4.3: Using float() for Decimals

````python
# Using float() for decimal numbers
price = float(input("Enter the price: "))
tax_rate = 0.06
total = price + (price * tax_rate)
print("Total with tax:", total)
````

## 5️⃣ Understanding Syntax Errors

Syntax errors occur when Python doesn't understand your code because it doesn't follow Python's rules (syntax).

Common syntax errors include:
- Misspelled variable names
- Missing parentheses or quotation marks
- Incorrect indentation
- Incomplete expressions

### Example 5.1: Common Syntax Errors

**Run each cell below to see the error messages. Then fix the code and run it again!**

````python
# Error 1: Misspelled variable name
length = 44
print(lenth)  # This will cause a NameError - can you fix it?
````

````python
# Error 3: Incomplete expression
result = 3 +  # This will cause a SyntaxError - can you fix it?
````

````python
# Error 4: Missing parenthesis
print("Python is fun"  # This will cause a SyntaxError - can you fix it?
````

### Tips for Debugging:

1. **Read the error message carefully** - it tells you what's wrong and where!
2. **Check the line number** mentioned in the error
3. **Look for common mistakes**: missing quotes, parentheses, or misspellings
4. **Run your code frequently** to catch errors early
5. **Don't panic!** Errors are normal and help you learn
