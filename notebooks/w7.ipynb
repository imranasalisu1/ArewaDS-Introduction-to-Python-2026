The counting pattern: get(key, 0) + 1.

````python
text = "the cat and the hat and the bat"
counts = {}
for word in text.split():
    counts[word] = counts.get(word, 0) + 1
print(counts)
````

````python
counts = {}
for word in text.split():
    if word in counts:
        counts[word] = counts[word] + 1
    else:
        counts[word] = 1
print(counts)
````

### TO DO 3 (Level 2: Apply): above_average

Write `above_average(numbers)` that **returns a new list** with the values greater than the average. It must not change the list passed in.

````python
# TO DO 3
def above_average(numbers):
    """Returns a new list containing all numbers strictly greater than the average."""
    # Edge case: Return an empty list if the input list is empty
    if not numbers:
        return []

    # Calculate the average (mean)
    avg = sum(numbers) / len(numbers)

    # Filter numbers greater than the average into a new list
    result = [num for num in numbers if num > avg]

    return result
````

````python
# CHECK: run this cell. Do not change it.
data = [70, 85, 90, 55, 100]
assert above_average(data) == [85, 90, 100], "average is 80"
assert data == [70, 85, 90, 55, 100], "do not change the original list"
assert above_average([5, 5, 5]) == [], "nothing is above the average"
print("All checks passed")
````

### TO DO 4 (Level 2: Apply): tally

Write `tally(items)` that takes a list of strings and returns a dictionary counting how many times each one appears.

````python
def tally(items):
    """Takes a list of strings and returns a dictionary with the count of each string."""
    counts = {}
    for item in items:
        if item in counts:
            counts[item] += 1
        else:
            counts[item] = 1
    return counts
````

````python
# CHECK: run this cell. Do not change it.
assert tally(["yes", "no", "yes", "yes"]) == {"yes": 3, "no": 1}
assert tally([]) == {}
assert tally(["a"]) == {"a": 1}
print("All checks passed")
````

### TO DO 5 (Level 3: Extend): Averages per student from a file

Write `student_averages(filename)` that reads lines like `Ana,88` and returns a dictionary from each name to that student's **average** score, rounded to 1 decimal. Hint: first build a dictionary from name to a **list** of scores.

````python
def student_averages(filename):
    """Reads a file with 'Name,Score' lines and returns a dict mapping
    each name to their average score rounded to 1 decimal place.
    """
    scores_dict = {}

    # Step 1: Read the file and map each name to a list of scores
    with open(filename, 'r') as file:
        for line in file:
            line = line.strip()
            if not line:  # Skip empty lines
                continue

            name, score_str = line.split(',')
            score = float(score_str)

            # Append score to existing list or start a new list
            if name in scores_dict:
                scores_dict[name].append(score)
            else:
                scores_dict[name] = [score]

    # Step 2: Compute the average for each student rounded to 1 decimal place
    averages_dict = {}
    for name, scores in scores_dict.items():
        avg = sum(scores) / len(scores)
        averages_dict[name] = round(avg, 1)

    return averages_dict
````

````python
# CHECK: run this cell. Do not change it.
result = student_averages("grades.txt")
assert result == {"Ana": 87.0, "Ben": 76.0, "Chidi": 81.0}, f"got {result}"
print("All checks passed")
````

**Question 2.** Choose a list, a dictionary, or a tuple for each, and say why: (a) temperatures for each day of a week, (b) student ID to student name, (c) a (latitude, longitude) pair.

_Your answer:_ 
Here are the best data structure choices for each scenario, along with the reasoning:

---

### (a) Temperatures for each day of a week $\rightarrow$ **List**

* **Why:** Temperatures recorded over consecutive days are sequential, ordered, and subject to change.
* **Reasoning:** A **list** is ordered and **mutable**. Since temperatures are collected over time (Day 1 through Day 7), keeping a chronological order matters. Additionally, if a reading is updated or corrected later, a list allows for in-place modifications.
* **Example:** `temps = [32.5, 30.1, 28.0, 31.4, 33.0, 29.8, 30.5]`

---

### (b) Student ID to student name $\rightarrow$ **Dictionary**

* **Why:** This requires a direct mapping between a unique key and an associated value.
* **Reasoning:** A **dictionary** is designed for key-value store lookups. Student IDs are unique identifiers (keys), making it fast and efficient to retrieve, add, or update the corresponding student name (value) without iterating through an entire list.
* **Example:** `students = {1001: "Imran Salisu", 1002: "Fatima Bello"}`

---

### (c) A (latitude, longitude) pair $\rightarrow$ **Tuple**

* **Why:** Represents a fixed, small grouping of related values that shouldn't change.
* **Reasoning:** A **tuple** is **immutable**. A geographical coordinate consists of a fixed structure where order matters (latitude first, longitude second) and the values together represent a single compound point that should not be accidentally altered. Tuples are also lighter in memory than lists.
* **Example:** `location = (12.1628, 6.6614)`
